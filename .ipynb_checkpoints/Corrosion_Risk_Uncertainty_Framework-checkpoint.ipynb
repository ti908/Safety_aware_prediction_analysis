
# Uncertainty-Aware Corrosion Risk Assessment for FPSO Pipelines

This notebook implements:
- Baseline: Distance-based uncertainty
- Advanced: Bayesian GMM & Ensemble uncertainty
- Comparative evaluation

Dataset: PRDS_Merged_Docs.xlsx


```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("PRDS_Merged_Docs.xlsx")
df.head()
```

## Feature Engineering

```

df['Severity'] = 1 - (df['RWT'] / df['Original_Thickness'])
df['Corrosion_Rate_proxy'] = df['Original_Thickness'] - df['RWT']
df['RUL_proxy'] = df['RWT'] / (df['Corrosion_Rate_proxy'] + 1e-6)
```

```

features = ['Severity','Corrosion_Rate_proxy','RUL_proxy','Temperature','Pressure']
X = df[features].fillna(df[features].mean())
```

## Baseline: Distance-Based Uncertainty

```

from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
df['Cluster'] = kmeans.fit_predict(X)

centers = kmeans.cluster_centers_

df['U_dist'] = np.linalg.norm(X - centers[df['Cluster']], axis=1)
df['Confidence_dist'] = 1 / (1 + df['U_dist'])
```

## Bayesian GMM Uncertainty

```

from sklearn.mixture import BayesianGaussianMixture

bgmm = BayesianGaussianMixture(n_components=5, random_state=42)
bgmm.fit(X)

df['U_BGMM'] = -bgmm.score_samples(X)
df['Confidence_BGMM'] = np.exp(-df['U_BGMM'])
```

## Ensemble Uncertainty

```

from sklearn.ensemble import RandomForestRegressor

models = [RandomForestRegressor(n_estimators=100, random_state=i) for i in range(10)]

preds = []
for m in models:
    m.fit(X, df['Severity'])
    preds.append(m.predict(X))

preds = np.array(preds)

df['U_ens'] = preds.std(axis=0)
df['Confidence_ens'] = 1 - df['U_ens']
```

## Risk Score Computation

```

df['Risk_dist'] = df['Severity'] * (1 / df['RUL_proxy']) * (1 - df['Confidence_dist'])
df['Risk_BGMM'] = df['Severity'] * (1 / df['RUL_proxy']) * (1 - df['Confidence_BGMM'])
df['Risk_ens'] = df['Severity'] * (1 / df['RUL_proxy']) * (1 - df['Confidence_ens'])
```

## Comparison Analysis

```

print(df[['Risk_dist','Risk_BGMM','Risk_ens','Severity']].corr())
```

```

sns.boxplot(data=df[['Risk_dist','Risk_BGMM','Risk_ens']])
plt.title("Risk Score Comparison")
plt.show()
```

```

sns.scatterplot(x=df['Severity'], y=df['Risk_BGMM'])
plt.title("Severity vs Bayesian Risk")
plt.show()
```
