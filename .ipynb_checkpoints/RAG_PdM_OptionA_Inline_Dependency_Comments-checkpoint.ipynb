This version mirrors the pipeline but adds inline comments explaining installations and fallbacks.

```python

import os, sys, json, math, re, random, sqlite3, subprocess, shutil, warnings
from pathlib import Path
warnings.filterwarnings("ignore")

RANDOM_STATE = 42
random.seed(RANDOM_STATE)

ROOT = Path.cwd()
DATA = ROOT / "data"
RAW = DATA / "raw"
CLEAN = DATA / "clean"
INDEX = DATA / "index"
DB_PATH = INDEX / "rag_pdms.sqlite"

for p in [DATA, RAW, CLEAN, INDEX]:
    p.mkdir(parents=True, exist_ok=True)

def safe_copy(src, dst):
    try:
        if src.exists() and not dst.exists():
            shutil.copyfile(src, dst)
            print(f"Copied: {src.name} -> {dst}")
    except Exception as e:
        print(f"[note] could not copy {src}: {e}")

safe_copy(Path("/mnt/data/Basics_of_Rotating_Industrial_Equipment.pdf"), RAW / "Basics_of_Rotating_Industrial_Equipment.pdf")
safe_copy(Path("/mnt/data/Engineers_Guide_to_Rotating_Equipment-The_Pocket_Reference.pdf"), RAW / "Engineers_Guide_to_Rotating_Equipment-The_Pocket_Reference.pdf")
safe_copy(Path("/mnt/data/MAINTENANCE_OF_MECHANICAL_AND_ELECTRICAL_EQUIPMENT.pdf"), RAW / "MAINTENANCE_OF_MECHANICAL_AND_ELECTRICAL_EQUIPMENT.pdf")
safe_copy(Path("/mnt/data/ai4i2020.csv"), DATA / "ai4i2020.csv")

def _pip(pkg, why):
    try:
        import importlib; importlib.import_module(pkg)
        print(f"[ok] {pkg} available — {why}")
    except Exception:
        try:
            print(f"[install] {pkg} — {why}")
            subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", pkg])
            print(f"[ok] installed {pkg}")
        except Exception as e:
            print(f"[warn] could not install {pkg}: {e}")

# Core scientific stack
_pip("pandas", "tabular data handling")
_pip("numpy", "numerics")
_pip("matplotlib", "plots")
_pip("scikit-learn", "PdM model + metrics")
_pip("imbalanced_learn", "SMOTE for imbalance")

# RAG stack
_pip("faiss-cpu", "ANN index for vector search")
_pip("sentence_transformers", "embeddings + reranker")
_pip("rank-bm25", "keyword retriever")
_pip("tiktoken", "token counting")
_pip("pypdf", "PDF parsing")

# Explainability
_pip("lime", "local explanations")
_pip("joblib", "serialization")
_pip("xgboost", "optional experiments")

def _try_import(name):
    try:
        __import__(name); return True
    except Exception as e:
        print(f"[note] {name} not available: {e}"); return False

HAVE_FAISS = _try_import("faiss")
HAVE_ST = _try_import("sentence_transformers")
HAVE_BM25 = _try_import("rank_bm25")
HAVE_TIKTOKEN = _try_import("tiktoken")
HAVE_PYPDF = _try_import("pypdf")
HAVE_LIME = _try_import("lime")

print("Capabilities ->", f"FAISS={HAVE_FAISS}, ST={HAVE_ST}, BM25={HAVE_BM25}, TIKTOKEN={HAVE_TIKTOKEN}, PYPDF={HAVE_PYPDF}, LIME={HAVE_LIME}")
print("Folders ready:", DATA, RAW, CLEAN, INDEX, sep="\n- ")
```

```python

# PdM model + LIME
import pandas as pd, numpy as np, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.calibration import CalibratedClassifierCV
from sklearn.metrics import classification_report, roc_auc_score, average_precision_score
from imblearn.over_sampling import SMOTE
from pathlib import Path

CSV_PATH = Path("data/ai4i2020.csv")
if not CSV_PATH.exists():
    raise FileNotFoundError("ai4i2020.csv not found in data/.")

df_raw = pd.read_csv(CSV_PATH)

COLS = {
    "air_temperature": None, "process_temperature": None, "rotational_speed": None,
    "torque": None, "tool_wear": None, "label": None
}
aliases = {
    "air_temperature": ["air temperature","air_temperature"],
    "process_temperature": ["process temperature","process_temperature"],
    "rotational_speed": ["rotational speed","rotational_speed","rpm"],
    "torque": ["torque","[nm]"],
    "tool_wear": ["tool wear","tool_wear","[min]"],
    "label": ["machine_failure","failure","target","label"]
}
for k, alts in aliases.items():
    for c in df_raw.columns:
        if any(a in c.lower() for a in alts): COLS[k]=c; break
if COLS["label"] is None:
    small = [c for c in df_raw.columns if df_raw[c].nunique()<=5 and pd.api.types.is_integer_dtype(df_raw[c])]
    COLS["label"] = small[0] if small else None

df = df_raw.copy()
df["temp_diff"] = df[COLS['process_temperature']] - df[COLS['air_temperature']]
df["rot_rads"] = (df[COLS['rotational_speed']] * 2*np.pi)/60.0
df["power_W"] = df[COLS['torque']] * df["rot_rads"]
df["overstrain_metric"] = df[COLS['torque']] * df[COLS['tool_wear']]

LABEL = COLS['label']
X = df.drop(columns=[LABEL], errors="ignore")
y = df[LABEL].astype(int)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
std_cols = [c for c in ["air_temperature","process_temperature","temp_diff","rotational_speed","torque","tool_wear","power_W","overstrain_metric"] if c in X_train.columns]
scaler = StandardScaler()
X_train.loc[:, std_cols] = scaler.fit_transform(X_train[std_cols])
X_test.loc[:, std_cols] = scaler.transform(X_test[std_cols])

sm = SMOTE(random_state=42)
X_tr, y_tr = sm.fit_resample(X_train, y_train)

rf = RandomForestClassifier(n_estimators=300, class_weight="balanced", random_state=42).fit(X_tr, y_tr)
cal_rf = CalibratedClassifierCV(rf, cv=3, method="isotonic").fit(X_tr, y_tr)

probs = cal_rf.predict_proba(X_test)[:,1]
print(classification_report(y_test, (probs>=0.5).astype(int)))
print("ROC AUC:", roc_auc_score(y_test, probs), " AP:", average_precision_score(y_test, probs))

try:
    from lime.lime_tabular import LimeTabularExplainer
    lime_explainer = LimeTabularExplainer(
        training_data=np.array(X_tr),
        feature_names=list(X_train.columns),
        class_names=["normal","failure"],
        discretize_continuous=True,
        random_state=42
    )
    print("[ok] LIME ready")
except Exception as e:
    lime_explainer = None
    print("[note] LIME not available:", e)
```

```python

# Minimal ingestion + retrieval + prompt simulation
import sqlite3, json, re, datetime
conn = sqlite3.connect(DB_PATH)
cur = conn.cursor()
cur.execute("CREATE TABLE IF NOT EXISTS chunks (id INTEGER PRIMARY KEY, manual_id TEXT, title TEXT, section TEXT, page_start INT, page_end INT, text TEXT, metadata_json TEXT)")
conn.commit()

def approx_token_count(text):
    try:
        import tiktoken
        enc = tiktoken.get_encoding("cl100k_base")
        return len(enc.encode(text))
    except Exception:
        return int(len(text.split())*1.3)

def split_by_structure(paras, tmin=300, tmax=800, overlap=0.15):
    out, buf, tok = [], [], 0
    for para in paras:
        if not para.strip(): continue
        t = approx_token_count(para)
        if tok + t > tmax and tok >= tmin:
            out.append("\\n".join(buf).strip())
            tail = max(1,int(len(buf)*overlap)); buf = buf[-tail:]; tok = sum(approx_token_count(x) for x in buf)
        buf.append(para); tok += t
    if buf: out.append("\\n".join(buf).strip())
    return out

def ingest_pdf(pdf_path: Path):
    try:
        from pypdf import PdfReader
    except Exception:
        print("[note] pypdf not available; skipping:", pdf_path.name); return
    reader = PdfReader(str(pdf_path))
    pages = []
    for i, p in enumerate(reader.pages):
        try: pages.append((i+1, p.extract_text() or ""))
        except Exception: pages.append((i+1, ""))
    paras = []
    for pno, txt in pages:
        for ln in (txt or "").splitlines():
            l = ln.strip()
            if not l: continue
            if re.match(r"^(chapter|section|introduction)\\b|\\b(safety|maintenance|pump|compressor|bearing|vibration)\\b", l.lower()):
                paras.append("### "+l)
            else:
                paras.append(l)
    chunks = split_by_structure(paras, 300, 800, 0.15)
    created_at = datetime.datetime.utcnow().isoformat()
    title = pdf_path.stem; manual_id = title[:40].lower().replace(" ","_")
    for ch in chunks:
        section = None
        for ln in ch.splitlines():
            if ln.startswith("### "): section = ln.strip("# ").strip(); break
        meta = {"manual_id": manual_id, "created_at": created_at, "source_pdf": pdf_path.name}
        cur.execute("INSERT INTO chunks (manual_id,title,section,page_start,page_end,text,metadata_json) VALUES (?,?,?,?,?,?,?)",
                    (manual_id,title,section,1,1,ch,json.dumps(meta)))
    conn.commit()

for pdf in Path("data/raw").glob("*.pdf"):
    ingest_pdf(pdf)

import pandas as pd, numpy as np
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

rows = pd.read_sql_query("SELECT * FROM chunks", conn)
VEC = TfidfVectorizer(max_features=4096, ngram_range=(1,2))
DOCS = rows["text"].fillna("").tolist()
MAT = VEC.fit_transform(DOCS).astype(np.float32)
NN = NearestNeighbors(n_neighbors=10, metric="cosine").fit(MAT)

def hybrid_search(query, top_k=8, keywords=None):
    import re
    qv = VEC.transform([query])
    dist, ind = NN.kneighbors(qv, n_neighbors=min(top_k*2, len(DOCS)))
    ids = ind[0].tolist(); scores = (1.0 - dist[0]).tolist()
    cand = rows.iloc[ids].copy(); cand["sim"] = scores
    if keywords:
        pat = re.compile("|".join(re.escape(k) for k in keywords), re.IGNORECASE)
        cand["sim"] += cand["text"].apply(lambda t: 0.1 if pat.search(t or "") else 0.0)
    return cand.sort_values("sim", ascending=False).head(top_k).reset_index(drop=True)

def pack_context(cdf, token_budget=1200):
    seen, packed, meta, total = set(), [], [], 0
    for _, r in cdf.iterrows():
        t = (r["text"] or "").strip()
        if not t or t in seen: continue
        need = min(approx_token_count(t), token_budget-total)
        if total + need > token_budget: break
        seen.add(t); total += need
        header = f"[{r['title']} :: {r.get('section') or 'General'} :: pp.{r.get('page_start')}-{r.get('page_end')}]"
        packed.append(header + "\\n" + t)
        meta.append({"id": int(r["id"]), "manual": r["title"], "section": r.get("section"), "pages": f"{r.get('page_start')}-{r.get('page_end']}"})
    return "\\n\\n---\\n\\n".join(packed), meta

dp = {
    "torque": 30.0,
    "rotational_speed": 1500.0,
    "tool_wear": 5.0,
    "temp_diff": 5.0
}
qry = f"torque={dp['torque']} rpm={dp['rotational_speed']} tool_wear={dp['tool_wear']} temp_diff={dp['temp_diff']}"
cand = hybrid_search(qry, top_k=8, keywords=["bearing","vibration"])
ctx, meta = pack_context(cand, token_budget=1200)
print(json.dumps({"datapoint": dp, "stage1": {"short_answer":"normal operation"}, "stage2": {"answer":"normal", "citations":""}, "meta": meta}, indent=2)[:1200], "...")
```
