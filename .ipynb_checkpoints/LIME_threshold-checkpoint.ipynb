```python
# ============================================================
# 1. Load dataset & focus on one failure mode (OSF)
# ============================================================

# Adjust the path to your file if needed
df = pd.read_csv("Data_file/ai4i2020.csv")

# Target: Overstrain Failure (OSF)
y = df["OSF"].astype(int)

# Features: use numeric process variables only (excluding IDs)
feature_cols = [
    "Air temperature [K]",
    "Process temperature [K]",
    "Rotational speed [rpm]",
    "Torque [Nm]",
    "Tool wear [min]",
]
X = df[feature_cols]

# Train-test split (stratified to preserve OSF proportion)
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42, stratify=y
)
```
